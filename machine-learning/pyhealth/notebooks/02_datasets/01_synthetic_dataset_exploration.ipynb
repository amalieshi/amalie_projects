# 01 — Synthetic Dataset Exploration

Explore the local synthetic EHR dataset using PyHealth's built-in statistics tools.
No credentials required — run `generate_synthetic_data.py` first.

```python
from pyhealth_enterprise.datasets.synthetic import SyntheticEHRDataset

ds = SyntheticEHRDataset()
ds.load()
ds.stat()   # PyHealth built-in: prints patient/visit/code statistics
```

```python
ds.info()   # PyHealth built-in: prints schema info
```

```python
# Visit distribution per patient
import pandas as pd
import matplotlib.pyplot as plt

visit_counts = [len(p.visits) for p in ds.dataset.patients.values()]
pd.Series(visit_counts).hist(bins=10)
plt.xlabel('Visits per patient')
plt.title('Visit Count Distribution')
plt.show()
```

```python
# ICD code frequency
from collections import Counter

all_codes = []
for patient in ds.dataset.patients.values():
    for visit in patient.visits:
        all_codes.extend([c for c in visit.get_code_list('conditions', 'ICD9CM')])

top_codes = Counter(all_codes).most_common(15)
pd.DataFrame(top_codes, columns=['ICD9_Code', 'Count']).set_index('ICD9_Code').plot.barh()
plt.title('Top 15 Diagnosis Codes')
plt.tight_layout()
plt.show()
```

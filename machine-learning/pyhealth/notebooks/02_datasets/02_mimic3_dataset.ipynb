# 02 — MIMIC-III Dataset

Load and explore MIMIC-III using PyHealth.

**Requires PhysioNet credentials.** See `docs/data_access_guide.md`.
Set `PHYSIONET_USERNAME` and `PHYSIONET_PASSWORD` in `.env`, and download MIMIC-III to `data/raw/mimic3/`.

```python
from pyhealth_enterprise.config import settings

if not settings.has_physionet_credentials():
    print('Warning: PhysioNet credentials not configured in .env')
    print('See docs/data_access_guide.md for setup instructions')

if not settings.MIMIC3_DATA_PATH.exists():
    print(f'MIMIC-III data not found at {settings.MIMIC3_DATA_PATH}')
    print('See docs/data_access_guide.md to download')
```

```python
from pyhealth.datasets import MIMIC3

# start with dev=True (1000 patients) for exploration
# remove dev=True for full dataset
dataset = MIMIC3(
    root=str(settings.MIMIC3_DATA_PATH),
    tables=['DIAGNOSES_ICD', 'PRESCRIPTIONS', 'PROCEDURES_ICD', 'ADMISSIONS'],
    code_mapping={'DIAGNOSES_ICD': 'ICD9CM', 'PRESCRIPTIONS': 'ATC'},
    dev=True,  # remove for full dataset
)
dataset.stat()
dataset.info()
```

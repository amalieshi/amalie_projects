```python
# Direct PyHealth OMOP usage
from pyhealth.datasets import OMOP
from pyhealth_enterprise.config import settings

dataset = OMOP(
    root=str(settings.OMOP_DATA_PATH),
    tables=['condition_occurrence', 'drug_exposure', 'measurement', 'procedure_occurrence'],
    code_mapping={'condition_occurrence': 'ICD10CM', 'drug_exposure': 'ATC'},
    dev=True,
)
dataset.stat()
```

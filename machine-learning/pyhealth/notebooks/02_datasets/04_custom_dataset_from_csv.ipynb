# 04 — Custom Dataset from CSV

Load any hospital CSV export into PyHealth using `SampleDataset`.
Demonstrates code mapping for non-standard vocabularies.

```python
from pyhealth.datasets import SampleDataset
from pyhealth_enterprise.config import settings

# SampleDataset expects CSVs with specific column names
# See data/samples/*.csv for the expected schema
dataset = SampleDataset(
    root=str(settings.SYNTHETIC_DATA_PATH),
    tables=['patients', 'visits', 'diagnoses', 'medications', 'labs'],
    code_mapping={'diagnoses': 'ICD9CM', 'medications': 'ATC'},
)
dataset.stat()
```

```python
# CrossMap: normalize ICD-9 to ICD-10
from pyhealth.medcode import CrossMap

icd9_to_icd10 = CrossMap('ICD9CM', 'ICD10CM')
ndc_to_atc = CrossMap('NDC', 'ATC')

# Example mappings
print('ICD-9 -> ICD-10:')
for code in ['428.0', '250.00', '401.9']:
    mapped = icd9_to_icd10.map(code)
    print(f'  {code} -> {mapped}')
```

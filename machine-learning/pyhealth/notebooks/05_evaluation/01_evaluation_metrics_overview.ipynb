# 01 — Evaluation Metrics Overview

PyHealth metrics for all clinical tasks:
- **Binary tasks** (readmission, mortality, LOS): PR-AUC, ROC-AUC, F1, accuracy
- **Multilabel tasks** (drug recommendation): Jaccard, PR-AUC, F1
- **DDI safety**: DDI rate (lower = safer)

```python
import numpy as np
from pyhealth_enterprise.evaluation.metrics import evaluate_binary_classifier, evaluate_drug_recommendation, compare_models

# Synthetic predictions for illustration
y_true_binary = np.array([0, 1, 0, 1, 1, 0, 1, 0, 1, 0])
y_prob_binary  = np.array([0.1, 0.9, 0.2, 0.8, 0.7, 0.3, 0.85, 0.15, 0.75, 0.25])

retain_metrics = evaluate_binary_classifier(y_true_binary, y_prob_binary)
transformer_metrics = evaluate_binary_classifier(y_true_binary, y_prob_binary + np.random.uniform(-0.05, 0.05, 10).clip(-0.1, 0.1))

df = compare_models({'RETAIN': retain_metrics, 'Transformer': transformer_metrics})
print(df.round(4))
```

```python
# Visualize PR curve
from sklearn.metrics import precision_recall_curve
import matplotlib.pyplot as plt

precision, recall, _ = precision_recall_curve(y_true_binary, y_prob_binary)
plt.plot(recall, precision)
plt.xlabel('Recall')
plt.ylabel('Precision')
plt.title('Precision-Recall Curve')
plt.grid(True)
plt.show()
```
